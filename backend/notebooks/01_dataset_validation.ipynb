```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
DATASET_PATH = "../scripts/data/generated/customer_support_dataset.csv"

df = pd.read_csv(DATASET_PATH)
```

```python
print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\nColumns")

for col in df.columns:
    print("-", col)

print("\nData Types")

print(df.dtypes)
```

```python
df.head(10)
```

```python
print("=" * 60)
print("MISSING VALUES")
print("=" * 60)

missing = df.isnull().sum()

print(missing)

print("\nTotal Missing Values :", missing.sum())
```

```python
print("=" * 60)
print("MISSING VALUES")
print("=" * 60)

missing = df.isnull().sum()

print(missing)

print("\nTotal Missing Values :", missing.sum())
```

```python
duplicates = df["ticket_id"].duplicated().sum()

print("Duplicate Ticket IDs :", duplicates)
```

```python
duplicates = df["customer_email"].duplicated().sum()

print("Duplicate Emails :", duplicates)
```

```python
empty = (df["description"].str.strip() == "").sum()

print("Empty Descriptions :", empty)
```

```python
length = df["description"].str.len()

print("=" * 60)
print("DESCRIPTION STATISTICS")
print("=" * 60)

print(f"Average Length : {length.mean():.2f}")

print(f"Minimum Length : {length.min()}")

print(f"Maximum Length : {length.max()}")
```

```python
category_counts = df["category"].value_counts()

print(category_counts)

category_counts.plot(kind="bar")

plt.title("Category Distribution")

plt.ylabel("Tickets")

plt.show()
```

```python
priority_counts = df["priority"].value_counts()

print(priority_counts)

priority_counts.plot(kind="bar")

plt.title("Priority Distribution")

plt.ylabel("Tickets")

plt.show()
```

```python
sentiment_counts = df["sentiment"].value_counts()

print(sentiment_counts)

sentiment_counts.plot(kind="bar")

plt.title("Sentiment Distribution")

plt.ylabel("Tickets")

plt.show()
```
