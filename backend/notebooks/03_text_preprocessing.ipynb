```python
import pandas as pd
import re
import string

import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("stopwords")
```

```python
DATASET_PATH = "../data/generated/customer_support_dataset.csv"

df = pd.read_csv(DATASET_PATH)

df.head()
```

```python
def lowercase(text):

    return text.lower()
```

```python
sample = df.loc[0, "description"]

print(sample)

print()

print(lowercase(sample))
```

```python
def remove_punctuation(text):

    return text.translate(
        str.maketrans("", "", string.punctuation)
    )
```

```python
sample = lowercase(df.loc[0, "description"])

print(remove_punctuation(sample))
```

```python
def remove_numbers(text):

    return re.sub(r"\d+", "", text)
```

```python
sample = remove_punctuation(sample)

print(remove_numbers(sample))
```

```python
def tokenize(text):

    return word_tokenize(text)
```

```python
sample = remove_numbers(sample)

tokens = tokenize(sample)

tokens
```

```python
STOP_WORDS = set(stopwords.words("english"))

def remove_stopwords(tokens):

    return [

        word

        for word in tokens

        if word not in STOP_WORDS

    ]
```

```python
filtered = remove_stopwords(tokens)

filtered
```

```python
lemmatizer = WordNetLemmatizer()

def lemmatize(tokens):

    return [

        lemmatizer.lemmatize(word)

        for word in tokens

    ]
```

```python
lemmas = lemmatize(filtered)

lemmas
```

```python
def preprocess(text):

    text = lowercase(text)

    text = remove_punctuation(text)

    text = remove_numbers(text)

    tokens = tokenize(text)

    tokens = remove_stopwords(tokens)

    tokens = lemmatize(tokens)

    return " ".join(tokens)
```

```python
sample = df.loc[0, "description"]

print("=" * 80)

print("Original")

print(sample)

print()

print("=" * 80)

print("Processed")

print(preprocess(sample))
```
